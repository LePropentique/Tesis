# DIAGONALIZACION DEL HAMILTONIANO
Estamos interesados en Diagonalizar el siguiente hamiltoniano:


$$H_{\gamma, h} = \sum_{i}^{} [(1 + \gamma) S^{x}_iS^{x}_{i+1} + (1-\gamma)S^{y}_iS^{y}_{i+1} + hS^{z}_i]$$

Aplicando los operadores escaleras 
$$a_i^{\dagger} = S^{x}_{i} + i\,S^{y}_{i} \quad a_i = S^{x}_{i} - i\,S^{y}_{i}
$$
Podemos despejar $S_x$ y $S_y$ en términos de los nuevos opperadores, y  sabiendo que los $\sigma$ de cada sitio cumplen el algebra de clifford, podemos despejar $S_z$ de igual manera.
Como los operadores de spin de sitios distintos conmutan, porque actuan sobre espacios de hilbert distintos dentro del producto tensorial, se puede ver que los operadores escaleras tambien, pues no mezclan sitios. En ese orden de ideas, estos nuevos operadores se comportan como fermiones en un mismo sitio, pero como bosones en sitios distintos. "No existe transformacion en ejes principales que a su vez sea canonica bajo este set de conmmutadores", luego necesitamos una transformacion que haga que en sitios distintos los operadores se comporten tambien como fermiones. Para esto utilizamos la transformacion de Jordan Wigner, que permite que sitios distintos anticonmuten. La demostracion d e que esto funciona y por qué funciona están en las notas (con BCH sale trivialmente). Así, se aplica:
$$
\begin{aligned}
c_i &\equiv \exp\!\left[\pi i \sum_{j=1}^{i-1} a_j^{\dagger} a_j\right] a_i, \\
c_i^{\dagger} &\equiv a_i^{\dagger} \exp\!\left[-\pi i \sum_{j=1}^{i-1} a_j^{\dagger} a_j\right].
\end{aligned}
$$
Como el operador numero asociado se conserva
$c_i^{\dagger} c_i = a_i^{\dagger} a_i $
Se puede despejar trivialmente las a's y reemplazar: 
$$
\begin{aligned}
a_i &= \exp\!\left[-\pi i \sum_{j=1}^{i-1} c_j^{\dagger} c_j\right] c_i, \\
a_i^{\dagger} &= c_i^{\dagger} \exp\!\left[\pi i \sum_{j=1}^{i-1} c_j^{\dagger} c_j\right].
\end{aligned}
$$
Inuitivamente, cuando hablamos de sitios distintos, sin perdida de generalidad, $i<j$, luego el modo $a_i$ aparecera en el termino exponencial de $c_j$, y al reorganizarlos aparecerá una fase de -1, que garantiza la anticonmutacion. 
Con esta transformación, nuestro hamiltoniano queda como

$$H = \frac{1}{2}\sum_{i=1}^{N-1}\Big[\big(c_i^{\dagger}c_{i+1} + \gamma\, c_i^{\dagger}c_{i+1}^{\dagger}\big) + \text{h.c.}\Big]
    - h\sum_{i=1}^{N}\Big(c_i^{\dagger}c_i - \tfrac{1}{2}\Big) $$

Considerando una cadena física cíclica, es decir, cerrada, la trasnformacion de jordan wigner para el termino $a_N^{\dagger} a_1$ produce un términio de fase extra (Que termina siendo un término de paridad, segun se demuestra en las notas), así que, en este caso, el hamiltoniano queda como:

$$
\begin{aligned}
H ={}& \frac{1}{2}\sum_{i=1}^{N}\Big[\big(c_i^{\dagger}c_{i+1} + \gamma\, c_i^{\dagger}c_{i+1}^{\dagger}\big) + \text{h.c.}\Big]
      - h\sum_{i=1}^{N}\Big(c_i^{\dagger}c_i - \tfrac{1}{2}\Big) \\
    &- \frac{1}{2}\Big[\big(c_N^{\dagger}c_1 + \gamma\, c_N^{\dagger}c_1^{\dagger}\big) + \text{h.c.}\Big]\big(e^{i\pi\mathfrak{N}} + 1\big),
\end{aligned}
$$

Para resolverlo, queremos encontrar un set de operadores {$\eta_k$, $\eta_k^{\dagger}$} que dejen el hamiltoniano de forma

$$ H = \sum_{i} \Lambda_k \eta_k \eta_k^{\dagger} +C $$

de donde se puede leer el espectro de energia de manera sencilla, y los $\eta_k$ permiten navegar el espacio de fock. Lo anterior se consigue haciendo las transformaciones de Bogolioubov-Valatin. Recalcar que cuando hay invarianza traslacional, el paper de mbeng et al (https://arxiv.org/pdf/2009.09208) da la solucion, donde la invarianza traslacional permite que, luego de hacer fourier, el hamiltoniano se descomponga en matrices 2x2, y se aplicque bogolioubov en cada una de ellas. Cuando no, se utilizará el método de solucion general dado en el paper de Lieb (Annals of physics: 16, 407-406 (1961)-  https://ui.adsabs.harvard.edu/abs/1961AnPhy..16..407L/abstract)


# Método Numérico General

Para free ends, el hamiltoniano es puramente cuadrático. Sabemos que Para un hamiltoniano de la forma
$$
H = \sum_{i,j} \left[ c_i^{\dagger} A_{ij}\, c_j + \tfrac{1}{2}\left( c_i^{\dagger} B_{ij}\, c_j^{\dagger} + \text{h.c.} \right) \right] + \frac{Nh}{2}
$$

Se propone la siguiente transformacion
$$
\begin{aligned}
\eta_k &= \sum_i \left( g_{ki}\, c_i + h_{ki}\, c_i^{\dagger} \right), \\
\eta_k^{\dagger} &= \sum_i \left( g_{ki}\, c_i^{\dagger} + h_{ki}\, c_i \right)
\end{aligned}
$$
Para que quede diagonalizado, se debe cumplir que
$$
[\eta_k, H] - \Lambda_k \eta_k = 0
$$
Reemplazando nuestras ecuaciones de $\eta_k$, $\eta_k^{\dagger}$ en la anterior, se llega a que se debe cumplir el siguiente set de ecuaciones
$$
\begin{aligned}
\Lambda_k\, g_{ki} &= \sum_j \left( g_{kj} A_{ji} - h_{kj} B_{ji} \right), \\
\Lambda_k\, h_{ki} &= \sum_j \left( g_{kj} B_{ji} - h_{kj} A_{ji} \right)
\end{aligned}
$$
Haciendo la siguiente asignacion 
$$
\phi_{ki} = g_{ki} + h_{ki}, \qquad \psi_{ki} = g_{ki} - h_{ki}
$$
Nuestras ecuaciones sobre los $\Lambda_k$ quedan como
$$
\boldsymbol{\phi}_k (A - B) = \Lambda_k \boldsymbol{\psi}_k, \qquad
\boldsymbol{\psi}_k (A + B) = \Lambda_k \boldsymbol{\phi}_k
$$
Multiplicando cada ecuacion por $\Lambda_k$ y reemplazando:
$$
\boldsymbol{\phi}_k (A - B)(A + B) = \Lambda_k^2\, \boldsymbol{\phi}_k, \qquad
\boldsymbol{\psi}_k (A + B)(A - B) = \Lambda_k^2\, \boldsymbol{\psi}_k
$$

**Caso $\Lambda_k \neq 0$:**

$$
\boldsymbol{\psi}_k = \frac{1}{\Lambda_k}\, \boldsymbol{\phi}_k (A - B)
$$

**Caso $\Lambda_k = 0$:**

$$
\boldsymbol{\phi}_k (A - B) = 0, \qquad \boldsymbol{\psi}_k (A + B) = 0
$$

$$
\boldsymbol{\phi}_k \cdot \boldsymbol{\phi}_{k'} = \delta_{kk'}, \qquad
\boldsymbol{\psi}_k \cdot \boldsymbol{\psi}_{k'} = \delta_{kk'}
$$

$$
H = \sum_k \Lambda_k\, \eta_k^{\dagger} \eta_k + \frac{1}{2}\left( \sum_i A_{ii} - \sum_k \Lambda_k \right) + \frac{Nh}{2},
\qquad
E_0 = -\frac{1}{2} \sum_k \Lambda_k
$$

```python
#Construye las matrices segun campo transverso y el término de anisotropia
import numpy as np
def build_AB(N, gamma, h):
    """Matrices A (simétrica) y B (antisimétrica) con extremos libres."""
    A = -h * np.eye(N) + 0.5 * (np.eye(N, k=1) + np.eye(N, k=-1))
    B = 0.5 * gamma * (np.eye(N, k=1) - np.eye(N, k=-1))
    return A, B

#Resuelve la ecuacion de valares propios para Phi y Psi
def solve_phi_psi(A, B, tol=1e-7):
    """
    phi_k (A-B)(A+B) = Lambda_k^2 phi_k
    psi_k = phi_k (A-B) / Lambda_k      si Lambda_k != 0
    psi_k (A+B)(A-B) = 0                si Lambda_k == 0
    """
    M = A - B
 
    w, V = np.linalg.eigh(M @ M.T)          # (A-B)(A+B), autovectores en columnas
    Lam = np.sqrt(np.clip(w, 0.0, None))
    phi = V.T                               # filas
    psi = np.zeros_like(phi)
 
    zero = Lam <= tol
 
    # Caso Lambda_k != 0
    psi[~zero] = (phi[~zero] @ M) / Lam[~zero, None]
 
    # Caso Lambda_k = 0
    n0 = zero.sum()
    if n0 > 0:
        w2, V2 = np.linalg.eigh(M.T @ M)    # (A+B)(A-B)
        psi[zero] = V2[:, :n0].T
        Lam[zero] = 0.0
 
    return Lam, phi, psi

#Devuelve los coeficientes de eta:
def eta_coefficients(phi, psi):
    """eta_k = sum_i ( g_ki c_i + h_ki c_i^† ),  T = [[g, h], [h, g]]."""
    g = 0.5 * (phi + psi)
    h = 0.5 * (phi - psi)
    T = np.block([[g, h], [h, g]])
    return g, h, T
```

Este simple codigo permite la diagonalizacion, pero no a la precision que nosotros queremos. Como nuestro objetivo es visualizar como se correlacionan los majorana, en ultimas, tendrémos que ver como se comportan la correlacion entre dos sitios lejanos. Esta correlacion, para un sistema con simetria traslacional, decae exponencialmente (De Mbeng):


Por ello estamos interesados en trabajar con precision arbitraria. Para ello vamos a darnos cuenta de que este problema se reduce a un problema SVD antes que un problema de valores propios. Antes de todo, como, por estructura, A es simétrica y B antisimétrica, es claro que $(A+B)^T = (A-B)$. Tomando $M=A+B$ y considerando las matrices

$$
\phi =
\begin{pmatrix}
\text{---} & \boldsymbol{\phi}_1 & \text{---} \\
 & \vdots & \\
\text{---} & \boldsymbol{\phi}_k & \text{---} \\
 & \vdots & \\
\text{---} & \boldsymbol{\phi}_N & \text{---}
\end{pmatrix},
\qquad
\psi =
\begin{pmatrix}
\text{---} & \boldsymbol{\psi}_1 & \text{---} \\
 & \vdots & \\
\text{---} & \boldsymbol{\psi}_k & \text{---} \\
 & \vdots & \\
\text{---} & \boldsymbol{\psi}_N & \text{---}
\end{pmatrix}
\qquad
\Lambda =
\begin{pmatrix}
\Lambda_1 & & & \\
 & \Lambda_2 & & \\
 & & \ddots & \\
 & & & \Lambda_N
\end{pmatrix}

$$
Las ecuaciones
$$
\boldsymbol{\phi}_k (A - B) = \Lambda_k \boldsymbol{\psi}_k, \qquad
\boldsymbol{\psi}_k (A + B) = \Lambda_k \boldsymbol{\phi}_k
$$

quedan respectivamente:
$$
\phi M = \Lambda \psi \qquad
\psi M^T = \Lambda \phi
$$
La condicion para que sea una transformacion canonica se traduce en que $\phi$ y $\psi$ sean ortogonales. Asi, si despejamos M y $M^T$ respectivamente, obtenemos
$$
 M = \phi^T \Lambda \psi \qquad
 M^T = \psi^T \Lambda \phi
$$
M es real, y $\phi$ y$ $\psi$ son ortogonales,  y $\Lambda$ es diagonal, luego estamos ante la descomposicion de M en valores singulares. Esto es perfecto porque este procedimiento computacionalmente se puede hacer hasta precision arbitraria! Con esto en mente, podemos modificar nuestra codigo para que funcione con mpmath. Para N=200 el código está demorando 5 minutos





```python
import numpy as np
from mpmath import mp
 
mp.dps = 50
 
 
def build_AB(N, gamma, h):
    """Matrices A (simétrica) y B (antisimétrica) con extremos libres."""
    A = -h * np.eye(N) + 0.5 * (np.eye(N, k=1) + np.eye(N, k=-1))
    B = 0.5 * gamma * (np.eye(N, k=1) - np.eye(N, k=-1))
    return A, B

def solve_phi_psi(A, B):
    """
    (A-7) como SVD:  A - B = phi^T Lambda psi
        phi_k = filas de U^T,  psi_k = filas de V,  Lambda_k = valores singulares
    Devuelve Lam (vector), phi, psi (matrices N x N) en precisión mp.dps.
    """
    M = mp.matrix(A.tolist()) - mp.matrix(B.tolist())
    U, Lam, V = mp.svd_r(M)          # M = U diag(Lam) V,  Lam de mayor a menor
    phi = U.T
    psi = V
    return Lam, phi, psi
 
 
def eta_coefficients(phi, psi):
    """eta_k = sum_i ( g_ki c_i + h_ki c_i^† ),  T = [[g, h], [h, g]]."""
    N = phi.rows
    g = (phi + psi) / 2
    h = (phi - psi) / 2
    T = mp.zeros(2 * N, 2 * N)
    for i in range(N):
        for j in range(N):
            T[i, j] = T[i + N, j + N] = g[i, j]
            T[i, j + N] = T[i + N, j] = h[i, j]
    return g, h, T
```

## Con Majoranas

Definimos dos operadores de Majorana por sitio:

$$
a_j = c_j + c_j^{\dagger}, \qquad b_j = i\,(c_j - c_j^{\dagger}), \qquad c_j = \tfrac{1}{2}\,(a_j - i\, b_j),
$$

que son hermíticos y cumplen el álgebra de Clifford:

$$
a_j^{\dagger} = a_j, \quad b_j^{\dagger} = b_j, \qquad
\{a_i, a_j\} = \{b_i, b_j\} = 2\delta_{ij}, \qquad \{a_i, b_j\} = 0 .
$$

Con $\boldsymbol{\gamma} = (a_1, \dots, a_N, b_1, \dots, b_N)^{T}$, todo Hamiltoniano cuadrático se escribe como

$$
H = \frac{i}{4} \sum_{\alpha, \beta} h_{\alpha\beta}\, \gamma_\alpha \gamma_\beta + \text{cte},
$$

donde la matriz $h$ es real y antisimétrica por construcción:

- Antisimétrica: como $\gamma_\alpha \gamma_\beta = -\gamma_\beta \gamma_\alpha$ para $\alpha \neq \beta$ y $\gamma_\alpha^2 = 1$, la parte simétrica de $h$ solo aporta una constante.
- Real: como $(i\gamma_\alpha\gamma_\beta)^{\dagger} = i\gamma_\alpha\gamma_\beta$, la condición $H^{\dagger} = H$ exige $h^{*} = h$.

Sustituyendo $c_j = \tfrac{1}{2}(a_j - i b_j)$ en el Hamiltoniano de extremos libres (la constante $\tfrac{1}{2}\operatorname{Tr}A + \tfrac{Nh}{2}$ se anula):

$$
H = -\frac{i}{2} \sum_{i,j} a_i\, M_{ij}\, b_j,
\qquad M = A - B,
\qquad
h =
\begin{pmatrix}
0 & -M \\
M^{T} & 0
\end{pmatrix}.
$$

Solo aparecen acoplamientos $a$–$b$.

Ahora, para toda matriz real antisimétrica $h$ de $2N \times 2N$ existe una matriz ortogonal real $O$ tal que

$$
O\, h\, O^{T} = \bigoplus_{k=1}^{N}
\begin{pmatrix}
0 & -\varepsilon_k \\
\varepsilon_k & 0
\end{pmatrix},
\qquad \varepsilon_k \geq 0 .
$$

La transformación debe ser real y ortogonal para que los nuevos operadores $\boldsymbol{\gamma}' = O\boldsymbol{\gamma}$ sigan siendo Majoranas: real para que sean hermíticos, y ortogonal para que se preserve $\{\gamma'_\alpha, \gamma'_\beta\} = 2\delta_{\alpha\beta}$ (transformación canónica). Una matriz unitaria compleja también diagonaliza $ih$, pero los operadores resultantes no son hermíticos; y una transformación real no ortogonal ($S h S^{T}$) puede simplificar más la matriz, pero rompe el álgebra.

*Podemos encontrar la relacion entre el método anterior y este de manera muy sencilla:*

$$
\boldsymbol{\eta} = T\,\mathbf{c} = T\,L'^{-1} L'\,\mathbf{c}
\quad\Longrightarrow\quad
L'\boldsymbol{\eta} = L'\,T\,L'^{-1}\, L'\mathbf{c}
$$

$$
\begin{pmatrix}
c_1 + c_1^{\dagger} \\ \vdots \\ c_N + c_N^{\dagger} \\ i\,(c_1 - c_1^{\dagger}) \\ \vdots \\ i\,(c_N - c_N^{\dagger})
\end{pmatrix}
=
\begin{pmatrix}
\mathbb{1} & \mathbb{1} \\
i\,\mathbb{1} & -i\,\mathbb{1}
\end{pmatrix}
\begin{pmatrix}
c_1 \\ \vdots \\ c_N \\ c_1^{\dagger} \\ \vdots \\ c_N^{\dagger}
\end{pmatrix},
\qquad
\begin{pmatrix}
c_1 + c_1^{\dagger} \\ \vdots \\ c_N + c_N^{\dagger} \\ i\,(c_1 - c_1^{\dagger}) \\ \vdots \\ i\,(c_N - c_N^{\dagger})
\end{pmatrix}
=
\begin{pmatrix}
a_1 \\ \vdots \\ a_N \\ b_1 \\ \vdots \\ b_N
\end{pmatrix}
= \boldsymbol{\gamma}
$$

$$
L' =
\begin{pmatrix}
\mathbb{1} & \mathbb{1} \\
i\,\mathbb{1} & -i\,\mathbb{1}
\end{pmatrix},
\qquad
L' L'^{\dagger} = 2\,\mathbb{1}
\quad (\text{unitaria salvo el factor } 2)
$$

$$
\boldsymbol{\gamma}' = \tilde{T}\,\boldsymbol{\gamma},
\qquad
\tilde{T} = L'\,T\,L'^{-1} =
\begin{pmatrix}
\phi & 0 \\
0 & \psi
\end{pmatrix}
\quad (\text{ortogonal})
$$

```python
def O_tilde(phi, psi):
    """
    Matriz ortogonal que conecta los Majoranas de los sitios con los de los modos:
        gamma' = O_tilde gamma,   O_tilde = L' T L'^{-1} = [[phi, 0], [0, psi]]
    con gamma = (a_1..a_N | b_1..b_N) y gamma' = (a'_1..a'_N | b'_1..b'_N).
    """
    N = phi.rows
    Ot = mp.zeros(2 * N, 2 * N)
    for i in range(N):
        for j in range(N):
            Ot[i, j] = phi[i, j]
            Ot[N + i, N + j] = psi[i, j]
    return Ot
```

# Método para Simetría Translacional

El operador de paridad fermiónica

$$
P = e^{i\pi\mathfrak{N}}, \qquad \mathfrak{N} = \sum_{j=1}^{N} c_j^{\dagger} c_j = \sum_{j=1}^{N} \left( S^{z}_j + \tfrac{1}{2} \right),
$$

conmuta con el Hamiltoniano, $[H, P] = 0$: cada término cambia $\mathfrak{N}$ en $0$ o en $\pm 2$. Por lo tanto la paridad se conserva, el espacio de Hilbert se separa en un sector par ($P = +1$) y uno impar ($P = -1$), y la evolución temporal nunca los mezcla.

En la práctica, en cada sector de paridad, el hamiltoniano es cuadrático y podemos usar el código inicial. Como primer método, usaremos el codigo anterior en cada sector de paridad y compararemos los niveles de energia, y aquel que sea el menor sera el ground state. Como segundo método, podemos usar el procedimiento encontrado en el paper de Mbeng, que es el más típico.

### Método 1
En la revision de la literatura, vimos que en toda la familia de parámetros, puede ser que el ground state del sistema viva en el par o en el impar, y el problema es que a veces en el impar hay vacios que no son físicos, luego el vacio asociado a dicho sector se toma como el primer estado escitado. En general, lo que e debe de hacer acá es calcular los ground states fisicos de cada sector y comparar (Tambien en el del metodo 2). Solo por ahora, se cogera la del sector par!. ### Método 2
 Procedimiento (Mbeng, Russomanno y Santoro)

1. En cada sector, $P$ es un número y el Hamiltoniano a-cíclico coincide con un Hamiltoniano cuadrático $H_s$, con $c_{N+1} = s\, c_1$ y $s = -P$.
2. $H_s$ es invariante por traslaciones, así que se pasa a momentos con una transformada de Fourier. El problema se separa en bloques independientes de $2\times 2$, uno por cada par $(k, -k)$.
3. Cada bloque se diagonaliza con una transformación de Bogoliubov.
4. De cada sector se conservan solo los estados con la paridad correcta. El estado fundamental es el de menor energía entre los dos sectores.

| sector | $P$ | $\mathfrak{N}$ | condición de contorno | momentos |
|---|---|---|---|---|
| $s = -1$ | $+1$ | par | antiperiódica (ABC), $c_{N+1} = -c_1$ | $k = \frac{2\pi}{N}\left(m + \tfrac{1}{2}\right)$ |
| $s = +1$ | $-1$ | impar | periódica (PBC), $c_{N+1} = c_1$ | $k = \frac{2\pi}{N}\, m$ |

con $m = 0, 1, \dots, N-1$ ($N$ par).

Fourier:

$$
c_j = \frac{1}{\sqrt{N}} \sum_k e^{ikj}\, c_k
$$

$$
H_s = \sum_k (\cos k - h)\, c_k^{\dagger} c_k
+ \frac{\gamma}{2} \sum_k \left( i \sin k \; c_k^{\dagger} c_{-k}^{\dagger} + \text{h.c.} \right)
+ \frac{Nh}{2}
$$

Bogoliubov para cada par $(k, -k)$, con $0 < k < \pi$

$$
\Psi_k =
\begin{pmatrix}
c_k \\ c_{-k}^{\dagger}
\end{pmatrix},
\qquad
\xi_k = \cos k - h,
$$

$$
H_k = \Psi_k^{\dagger}
\begin{pmatrix}
\xi_k & i\gamma \sin k \\
-i\gamma \sin k & -\xi_k
\end{pmatrix}
\Psi_k + \xi_k
= \Lambda_k \left( \eta_k^{\dagger}\eta_k + \eta_{-k}^{\dagger}\eta_{-k} - 1 \right) + \xi_k,
$$

$$
\Lambda_k = \sqrt{(\cos k - h)^2 + \gamma^2 \sin^2 k} .
$$

Modos sin pareja (solo en el sector PBC):

$k = 0$ y $k = \pi$ son su propio $-k$; no hay término de pares y quedan como

$$
(1 - h)\, n_0, \qquad -(1 + h)\, n_\pi .
$$

Energías:

$$
E^{\text{ABC}} = -\sum_{0<k<\pi} \Lambda_k + \sum_{0<k<\pi} \Lambda_k \left( n_k + n_{-k} \right)
$$

$$
E^{\text{PBC}} = h - \sum_{0<k<\pi} \Lambda_k + \sum_{0<k<\pi} \Lambda_k \left( n_k + n_{-k} \right) + (1 - h)\, n_0 - (1 + h)\, n_\pi
$$

Selección por paridad:

Cada excitación $\eta_k^{\dagger}$ cambia la paridad, y el vacío de los pares $(k, -k)$ es par.

- **ABC** (se exige $\mathfrak{N}$ par): el vacío de los $\eta$ es físico,
$$
E_0^{\text{ABC}} = -\sum_{0<k<\pi} \Lambda_k .
$$
- **PBC** (se exige $\mathfrak{N}$ impar): se toma la configuración de $n_0$, $n_\pi$ y excitaciones $\eta$ de menor energía con $n_0 + n_\pi + \#\eta$ impar. Para $|h| < 1$ es $n_\pi = 1$, $n_0 = 0$ y ningún $\eta$:
$$
E_0^{\text{PBC}} = -1 - \sum_{0<k<\pi} \Lambda_k \qquad (|h| < 1).
$$

# Matriz de Covarianza. 


El vacío de los modos está definido por $\eta_k |0\rangle = 0$ para todo $k$ (y entonces $\langle 0 | \eta_k^{\dagger} = 0$). En un valor esperado de dos operadores, el único término que no se anula es

$$
\langle 0 | \eta_k\, \eta_l^{\dagger} | 0 \rangle = \langle 0 | \left( \{\eta_k, \eta_l^{\dagger}\} - \eta_l^{\dagger} \eta_k \right) | 0 \rangle = \delta_{kl} .
$$

Con $a'_k = \eta_k + \eta_k^{\dagger}$ y $b'_k = i\,(\eta_k - \eta_k^{\dagger})$:

$$
\langle a'_k a'_l \rangle = \delta_{kl}, \qquad
\langle b'_k b'_l \rangle = \delta_{kl}, \qquad
\langle a'_k b'_l \rangle = -i\,\delta_{kl}, \qquad
\langle b'_k a'_l \rangle = i\,\delta_{kl} .
$$

**Matriz de covarianza en la base de los modos**

$$
M_{\alpha\beta} = \frac{1}{2i} \left\langle [\gamma_\alpha, \gamma_\beta] \right\rangle,
\qquad
\langle \gamma_\alpha \gamma_\beta \rangle = \delta_{\alpha\beta} + i\, M_{\alpha\beta} .
$$

En el orden $\boldsymbol{\gamma}' = (a'_1, \dots, a'_N \,|\, b'_1, \dots, b'_N)$, cada modo solo está correlacionado consigo mismo:

$$
\langle -i\, a'_k b'_k \rangle = -1, \qquad
M' =
\begin{pmatrix}
0 & -\mathbb{1} \\
\mathbb{1} & 0
\end{pmatrix} .
$$

**Matriz de covarianza en la base de los sitios**

Como $\boldsymbol{\gamma}' = \tilde{O}\, \boldsymbol{\gamma}$, es decir $\gamma_\alpha = \sum_k \tilde{O}_{k\alpha}\, \gamma'_k$:

$$
M = \tilde{O}^{T} M' \tilde{O}
=
\begin{pmatrix}
0 & -\phi^{T} \psi \\
\psi^{T} \phi & 0
\end{pmatrix},
\qquad
\langle -i\, a_i b_j \rangle = -\sum_k \phi_{ki}\, \psi_{kj} .
$$

Cada Majorana de un sitio es una combinación de los Majoranas de muchos modos, de modo que aparecen correlaciones entre sitios distintos:

- $i = j$: magnetización, $\langle \sigma^z_i \rangle = M_{a_i b_i}$;
- $j = i \pm 1$: correlaciones de primeros vecinos ($\sigma^x\sigma^x$, $\sigma^y\sigma^y$);
- $|i - j|$ grande: decaimiento de las correlaciones con la distancia.

Los bloques $aa$ y $bb$ siguen siendo cero, porque $\tilde{O}$ no mezcla los $a$ con los $b$, y $M^2 = -\mathbb{1}$ (estado puro).

# Funciones de modo

Dada una ortogonal $O$ que lleva unos Majoranas a otros, $\boldsymbol{\gamma}' = O\,\boldsymbol{\gamma}$, cada Majorana nuevo es

$$
\gamma'_r = \sum_i O_{ri}\, \gamma_i ,
$$

y su **función de modo** es la fila $r$ de $O$, vista como función del índice de los Majoranas viejos:

$$
F(\gamma'_r)_i = O_{ri}, \qquad \sum_i F(\gamma'_r)_i^2 = 1 .
$$

Cada **modo físico** $m$ está formado por dos Majoranas emparejados, $\gamma'_\alpha = a'_m$ y $\gamma'_\beta = b'_m$ (en el orden por bloques, filas $\alpha = m$ y $\beta = n + m$ de $O$), así que tiene **dos** funciones de modo, $F(\gamma'_\alpha)$ y $F(\gamma'_\beta)$. Esto vale tanto para la ortogonal global $\tilde{O}$ (modos del Hamiltoniano) como para las ortogonales locales $O_{\mathcal{R}}$ (modos de Williamson de cada región).

**Una sola función por modo físico**

Las funciones $F(\gamma'_\alpha)$ y $F(\gamma'_\beta)$ no son únicas: una rotación del par $(\gamma'_\alpha, \gamma'_\beta)$ (fase del modo) las mezcla sin cambiar el modo físico, y lo mismo ocurre con una rotación del par $(a_i, b_i)$ de un sitio (fase de $c_i$). Para describir dónde vive cada modo se define una única función, que no depende de esas elecciones, sumando los cuadrados de las dos funciones sobre los dos Majoranas de cada sitio:

$$
F_m(i) = \frac{1}{2} \left[ F(\gamma'_\alpha)_{a_i}^2 + F(\gamma'_\alpha)_{b_i}^2 + F(\gamma'_\beta)_{a_i}^2 + F(\gamma'_\beta)_{b_i}^2 \right],
\qquad \sum_i F_m(i) = 1 .
$$

Con las ortogonales del problema, que no mezclan $a$ con $b$, $F(\gamma'_\alpha)$ solo tiene peso sobre los $a_i$ y $F(\gamma'_\beta)$ solo sobre los $b_i$, y queda

$$
F_m(i) = \frac{1}{2} \left[ F(\gamma'_\alpha)_{a_i}^2 + F(\gamma'_\beta)_{b_i}^2 \right].
$$

$F_m(i)$ es el peso del modo físico $m$ en el sitio $i$.

**Qué queremos estudiar**

Siguiendo la idea de las particiones:

1. **Sin partición:** funciones de modo de los modos globales (filas de $\tilde{O}$).
2. **Bipartición:** en cada mitad, el modo de Williamson más entrelazado ($\lambda_m \approx 0$) tiene peso en los dos extremos de la región. Los picos de ambas mitades junto al corte están emparejados entre sí y, al considerar las dos mitades juntas, se "cancelan", recuperando la estructura sin partición.
3. **Tripartición $\mathcal{A}|\mathcal{B}|\mathcal{C}$:** comparar las funciones $F_m$ de los modos de $\mathcal{B}$, de $\mathcal{C}$ y de $\mathcal{BC}$ como una sola región, para ver si los picos en el corte $\mathcal{B}|\mathcal{C}$ se cancelan y reproducen los modos de $\mathcal{BC}$, y qué modos de $\mathcal{B}$ quedan acoplados a la vez con $\mathcal{A}$ y con $\mathcal{C}$.
